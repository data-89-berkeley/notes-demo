# Demo Testing

One demo, *Surfaces, Level Sets and Gradients*, embedded three ways so they can be compared. The first two show the standalone HTML page (`demos/level-sets-gradients.html`) as it was written. The third is the same demo rebuilt as a book widget.

| | [1. Static HTML file](#demo-testing-static) | [2. HTML bundled into a widget](#demo-testing-bundled) | [3. Native widget](#demo-testing-native) |
|---|---|---|---|
| Shows in `jupyter book start` | No ("Document Not Found" in the frame) | Yes | Yes |
| Shows on the built site | Yes | Yes | Yes |
| Follows the book's theme (dark mode, fonts) | No | No | Yes |
| Covered by `npm test` | No | No | Yes |
| Work per demo | One line in `myst.yml` + an `{iframe}` cell | A three-line wrapper file | Port the page's JavaScript (hours) |

**2 versus 3.** In 2 the HTML file *is* the demo and the widget is only a box that displays it. In 3 the widget *is* the demo: the file's JavaScript was moved into the widget's source, so the book can style it, share plotly with it, and test it. If a page looks right as is, 2 is enough; 3 is for demos that should look like the rest of the book.
