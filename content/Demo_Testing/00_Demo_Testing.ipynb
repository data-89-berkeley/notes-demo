# Demo Testing

## 1. Static HTML file

- The file is listed under `static_files` in `myst.yml`, which copies it into the site's `build/` folder, and this cell shows that copy with the `{iframe}` directive (`style.css` gives the frame its height).
- Only the built site serves that folder, so in `jupyter book start` this frame shows the site's "Document Not Found" page. The page keeps its own styling and has no tests.

```{iframe} /notes-demo/build/level-sets-gradients.html
:width: 100%
:class: demo-frame col-page-right
:title: Surfaces, level sets and gradients (static file)
```

## 2. HTML bundled into a widget

- A three-line wrapper, `demos/src/demos/lsg_page.js`, imports the HTML file as text; the demo build bundles it into a widget that shows it in a frame sized to the page.
- Works in `jupyter book start` and on the built site with nothing to host. It is still the page's own code in a box: no dark mode, nothing to test.

```{anywidget} ../_widgets/lsg_page.js
:class: col-page-right
{}
```

## 3. Native widget

- The page's JavaScript rewritten in the book's widget format, `demos/src/demos/lsg_explorer.js`: same math and plots, but the controls and plots are built directly in the page, with no frame.
- Follows the book's theme, shares plotly with the page, and has unit tests. The cell's JSON can pick the opening tab (`"v1"` cross-sections, `"v2"` level sets, `"v3"` gradient field) and surface.

```{anywidget} ../_widgets/lsg_explorer.js
:class: col-page-right
{"tab": "v2"}
```

## Comparison

| | 1. Static HTML file | 2. HTML bundled into a widget | 3. Native widget |
|---|---|---|---|
| Shows in `jupyter book start` | No ("Document Not Found" in the frame) | Yes | Yes |
| Shows on the built site | Yes | Yes | Yes |
| Follows the book's theme (dark mode, fonts) | No | No | Yes |
| Covered by `npm test` | No | No | Yes |
| Work per demo | One line in `myst.yml` + an `{iframe}` cell | A three-line wrapper file | Port the page's JavaScript (hours) |

**2 versus 3.** In 2 the HTML file *is* the demo and the widget is only a box that displays it. In 3 the widget *is* the demo: the file's JavaScript was moved into the widget's source, so the book can style it, share plotly with it, and test it. If a page looks right as is, 2 is enough; 3 is for demos that should look like the rest of the book.
